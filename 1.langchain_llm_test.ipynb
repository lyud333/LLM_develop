```python
from dotenv import load_dotenv
from langchain_anthropic import ChatAnthropic

load_dotenv()

llm = ChatAnthropic(
    model="claude-sonnet-5",
    max_tokens=1024,
)
```

```python
response = llm.invoke("발로란트 챔피언스 상하이 주제곡 가사를 짜줘")
print(response.content)
```
