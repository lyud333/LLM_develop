```python
from dotenv import load_dotenv
from langchain_anthropic import ChatAnthropic

load_dotenv()

llm = ChatAnthropic(
    model="claude-sonnet-5",
    max_tokens=1024,
)

llm.invoke("배고픈데?, 뭐 먹을지 음식추천 5개해줘")
```

```python
response = llm.invoke("안녕? 파이썬 가상환경이 뭔지 3줄로 설명해줘")
print(response.content)
```
